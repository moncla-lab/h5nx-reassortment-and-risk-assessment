```python
import os
import pandas as pd
from Bio import SeqIO
```

```python
fasta_dir = '../2344b-lpai-2020-genome/results'

genes = ['pb2', 'pb1', 'pa', 'ha', 'np', 'na', 'mp', 'ns']

records = {}

for gene in genes:
    fasta = os.path.join(fasta_dir, f'aligned_{gene}.fasta')
    for record in SeqIO.parse(fasta, 'fasta'):
        if record.id not in records:
            records[record.id] = record
        else:
            records[record.id].seq += record.seq
records = list(records.values())
concatenated_genomes = os.path.join(fasta_dir, 'aligned_genome.fasta')
SeqIO.write(records, concatenated_genomes, 'fasta')
```
