```python
import pandas as pd
from datetime import datetime

def generate_strain_dates(build_dir):
    strains = f'{build_dir}/results/filtered_ha_filtered.txt'
    strain_dates = f'{build_dir}/results/strain_dates.csv'
    
    d = {'strain':[],'date':[]}
    
    with open(strains) as f:
        for l in f.read().split('\n'):
            l = l.strip()
            if l:
                d['strain'].append(l)
                date = l.split('|')[1]
                d['date'].append(date)
                # d['date'].append(datetime.strptime(l.split('|')[1], '%Y-%m-%d'))
    
    df = pd.DataFrame(d)
    
    # df['date'].apply(lambda date:date.year + ((date.month - 1) * 30 + date.day) / 365.0)
    
    df.to_csv(strain_dates)
```

```python
build_dirs = [
    '../2344b-lpai-2020-reannotated-noRefine',
    '../2321-2020-reannotated-noRefine'
]

for build_dir in build_dirs:
    generate_strain_dates(build_dir)
```
