```python
import pandas as pd
```

```python
metadata = '/Users/jort/coding/h5nx-reassortment-and-risk-assessment/data/h5nx_metadata.tsv'
```

```python
df = pd.read_csv(metadata, sep='\t')
df.head()
```

```python
txt = '/Users/jort/coding/h5nx-reassortment-and-risk-assessment/2344b-lpai-2020/config/dropped_strains-no-dates.txt'
outtxt = '/Users/jort/coding/h5nx-reassortment-and-risk-assessment/2344b-lpai-2020/config/dropped_strains.txt'

delims = [' ', ',', '\t']

dates = dict(zip(df['strain'], df['date']))

mapping = {}

out = []

with open(txt) as f:
    for l in f.read().split('\n'):
        if not l.startswith('#') and l.strip():
            for delim in delims:
                l = l.split(delim)[0]
            d = dates.get(l)
            if d:
                out.append(f'{l}|{d}')

with open(outtxt, 'w') as f:
    f.write('\n'.join(out))
```
