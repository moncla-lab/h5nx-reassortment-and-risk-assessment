```python
import pandas as pd
from collections import defaultdict
```

```python
infile = '/Users/jort/coding/h5nx-reassortment-and-risk-assessment/data/h5nx_metadata_dates.tsv'
outfile = '/Users/jort/coding/h5nx-reassortment-and-risk-assessment/data/h5nx_metadata_dates_filters.tsv'

df = pd.read_csv(infile, sep='\t')

# 2.3.4.4b or lpai filter

mapping = defaultdict(lambda:'F')

for k in ['2.3.4.4b', 'Am-nonGsGD', 'EA-nonGsGD']:
    mapping[k] = 'T'

df['2344b_or_lpai'] = df['h5_label_clade'].map(mapping)

# 2.3.2.1 filter

mapping = defaultdict(lambda:'F')

ls = ['', 'a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']
for k in [f'2.3.2.1{l}' for l in ls]:
    mapping[k] = 'T'

df['2321'] = df['h5_label_clade'].map(mapping)

df.to_csv(outfile, sep='\t', index=False)
```
