```python
import argparse
import pandas as pd

if __name__ == '__main__':
    parser = argparse.ArgumentParser()

    parser.add_argument('-m', '--metadata', action='store',
                        dest='metadata', required=True, help='Path to metadata tsv')
    parser.add_argument('-o', '--output_metadata', action='store',
                        dest='output_metadata', required=True, help='Path to metadata tsv')
    

    args = parser.parse_args()



    df = pd.read_csv(args.metadata, sep='\t')
    df['strain'] = '|'.join(df['strain'].str, df['date'].str)

    df.to_csv(args.output_metadata)
```
